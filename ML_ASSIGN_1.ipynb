```python
#7
import numpy as np

# Create a 3x3 matrix
matrix = np.arange(1, 10).reshape(3, 3)

# Extract the last row
print(matrix[-1])  # Selects the last row.

# Extract the last column
print(matrix[:, -1])  # Selects the last column.

# Extract the middle element
print(matrix[1, 1])  # Selects row 1, column 1.

# Extract the bottom-right 2x2 sub-matrix
print(matrix[1:, 1:])  # Selects the last 2 rows and last 2 columns.
```

```python
#9
import numpy as np

# Add a column array and a row array using broadcasting
result = np.array([[1], [2], [3]]) + np.array([10, 20, 30])

print(result)  # NumPy broadcasts both arrays and performs the addition.
```

```python
#10
import numpy as np

# Create an array of prices
prices = np.array([100, 200, 300])

# Add 10 to every price using broadcasting
result = prices + 10

print(result)  # The value 10 is automatically applied to every element.

# Broadcasting allows NumPy to perform operations between compatible arrays
# without writing a separate loop.
```

```python
#11
import numpy as np

# Create the marks array
marks = np.array([45, 67, 89, 32, 90, 55])

# Extract all failing marks below 40
failing = marks[marks < 40]
print(failing)  # Displays the marks below 40.

# Count students who scored above 80
count = (marks > 80).sum()
print(count)  # Counts the number of True values.

# Replace failing marks with 0
result = np.where(marks < 40, 0, marks)
print(result)  # Replaces marks below 40 with 0.
```

```python
#12
import numpy as np

# Create an array containing 12 elements
x = np.arange(12)

# Split the array into 4 equal parts
print(np.split(x, 4))  # 12 elements are divided into 4 parts.

# Try to split the array into 5 parts
try:
    print(np.split(x, 5))
except ValueError:
    print("Cannot split 12 elements into 5 equal parts.")

# Explanation:
# 12 cannot be divided equally into 5 parts.
```

```python
import numpy as np             # Import NumPy library
x = np.arange(12)              # Create an array from 0 to 11
print(x)                       # Display the original array
print(np.split(x, 3))          # Equal split into 3 parts
print(np.split(x, 4))          # Equal split into 4 parts
print(np.array_split(x, 5))    # Unequal split into 5 parts
```

```python
from google.colab import files
uploaded = files.upload()
from PIL import Image                          # Import Image from PIL
import matplotlib.pyplot as plt                # Import Matplotlib
import numpy as np                             # Import NumPy
img_array = np.array(Image.open("wallpaper.png"))  # Load image and convert it to an array
plt.imshow(img_array)                          # Display the image
plt.axis("off")                                # Hide the axes
plt.show()                                     # Show the image
```

```python
import matplotlib.pyplot as plt              # Import Matplotlib
height, width = img_array.shape[:2]          # Get the height and width of the image
start_y = (height - 200) // 2                # Calculate the starting row for the center
start_x = (width - 200) // 2                 # Calculate the starting column for the center
crop = img_array[start_y:start_y + 200,      # Crop 200 rows
                 start_x:start_x + 200]      # Crop 200 columns
plt.imshow(crop)                             # Display the cropped image
plt.axis("off")                              # Hide the axes
plt.show()                                   # Show the image
```

```python
import matplotlib.pyplot as plt              # Import Matplotlib
vertical_flip = img_array[::-1, :]           # Flip the image vertically
horizontal_flip = img_array[:, ::-1]         # Flip the image horizontally
plt.imshow(vertical_flip)                    # Display the vertical flip
plt.axis("off")                              # Hide the axes
plt.show()                                   # Show the image
plt.imshow(horizontal_flip)                  # Display the horizontal flip
plt.axis("off")                              # Hide the axes
plt.show()                                   # Show the image
```

```python
import matplotlib.pyplot as plt             # Import Matplotlib
red = img_array[:, :, 0]                    # Extract the Red channel
green = img_array[:, :, 1]                  # Extract the Green channel
blue = img_array[:, :, 2]                   # Extract the Blue channel
plt.imshow(red, cmap="gray")                # Display the Red channel
plt.title("Red Channel")                    # Add a title
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the image
plt.imshow(green, cmap="gray")              # Display the Green channel
plt.title("Green Channel")                  # Add a title
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the image
plt.imshow(blue, cmap="gray")               # Display the Blue channel
plt.title("Blue Channel")                   # Add a title
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the image
```

```python
import numpy as np                        # Import NumPy
average = np.mean(img_array)              # Calculate average brightness of the whole image
red = np.mean(img_array[:, :, 0])         # Calculate average brightness of Red channel
green = np.mean(img_array[:, :, 1])       # Calculate average brightness of Green channel
blue = np.mean(img_array[:, :, 2])        # Calculate average brightness of Blue channel
print("Average brightness:", average)     # Display overall average brightness
print("Red brightness:", red)             # Display Red channel average
print("Green brightness:", green)         # Display Green channel average
print("Blue brightness:", blue)           # Display Blue channel average
```

```python
import numpy as np                         # Import NumPy
import matplotlib.pyplot as plt             # Import Matplotlib
gray = np.mean(img_array, axis=2)           # Convert RGB image to grayscale by averaging R, G, B
binary = np.where(gray > 128, 255, 0)       # If brightness is above 128, make it white; otherwise black
plt.imshow(binary, cmap="gray")             # Display the black-and-white image
plt.axis("off")                             # Hide the axes
plt.show()                                  # Show the result
```

```python
import numpy as np                         # Import NumPy
import matplotlib.pyplot as plt             # Import Matplotlib

# 1. Image height, width, and number of channels
height, width, channels = img_array.shape   # Get image dimensions
print("Height:", height)                    # Display height
print("Width:", width)                      # Display width
print("Channels:", channels)                # Display number of channels


# 2. Average brightness
overall = np.mean(img_array)                # Calculate overall average brightness
red = np.mean(img_array[:, :, 0])           # Calculate average Red brightness
green = np.mean(img_array[:, :, 1])         # Calculate average Green brightness
blue = np.mean(img_array[:, :, 2])          # Calculate average Blue brightness

print("Overall brightness:", overall)       # Display overall brightness
print("Red brightness:", red)               # Display Red brightness
print("Green brightness:", green)           # Display Green brightness
print("Blue brightness:", blue)             # Display Blue brightness


# 3. Crop the center 200x200 region
start_y = (height - 200) // 2               # Find starting row
start_x = (width - 200) // 2                # Find starting column

crop = img_array[start_y:start_y + 200,
                 start_x:start_x + 200]     # Crop the center region

plt.imshow(crop)                            # Display cropped image
plt.title("Center Crop")
plt.axis("off")                             # Hide axes
plt.show()


# 4. Flip the image vertically and horizontally
vertical_flip = img_array[::-1, :]          # Flip image vertically
horizontal_flip = img_array[:, ::-1]        # Flip image horizontally

plt.imshow(vertical_flip)                   # Display vertical flip
plt.title("Vertical Flip")
plt.axis("off")
plt.show()

plt.imshow(horizontal_flip)                 # Display horizontal flip
plt.title("Horizontal Flip")
plt.axis("off")
plt.show()

# 5. Display RGB channels separately
red_channel = img_array[:, :, 0]            # Extract Red channel
green_channel = img_array[:, :, 1]          # Extract Green channel
blue_channel = img_array[:, :, 2]           # Extract Blue channel

plt.imshow(red_channel, cmap="gray")        # Display Red channel
plt.title("Red Channel")
plt.axis("off")
plt.show()

plt.imshow(green_channel, cmap="gray")      # Display Green channel
plt.title("Green Channel")
plt.axis("off")
plt.show()

plt.imshow(blue_channel, cmap="gray")       # Display Blue channel
plt.title("Blue Channel")
plt.axis("off")
plt.show()

# 6. Convert image to black and white
gray = np.mean(img_array, axis=2)            # Convert RGB image to grayscale
binary = np.where(gray > 128, 255, 0)       # Above 128 = white, otherwise black
plt.imshow(binary, cmap="gray")             # Display black-and-white image
plt.title("Thresholded Image")
plt.axis("off")
plt.show()
```
